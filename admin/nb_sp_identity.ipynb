## **Identity functions**

```python
# --------------------------------------------------------
#  Associate item identity function
# --------------------------------------------------------
def associate_item_identity(
        access_token: str
      , workspace_id: str
      , item_id: str
      , poll_timeout_seconds: int = 120
      , poll_interval_seconds: int = 5
) -> dict:
    """
    Assign the calling identity (SP, user, or managed identity) as the
    associated identity (default identity) of a Fabric item.

    This replaces the legacy owner-based dependency with the identity of
    the caller making the API request. Currently supported item types:
    Lakehouse and Eventstream.

    The API is asynchronous - a 202 Accepted response includes a Location
    header pointing to the long-running operation. This function polls that
    operation until it reaches Succeeded, Failed, or the timeout expires.

    :param access_token: OAuth2 bearer token from Entra ID.
    :param workspace_id: GUID of the Fabric workspace.
    :param item_id: GUID of the Fabric item to associate.
    :param poll_timeout_seconds: Maximum seconds to wait for the LRO.
    :param poll_interval_seconds: Seconds between polling attempts.
    :return: The final operation response body as a dict.
    """

    endpoint = (
        f"https://api.fabric.microsoft.com/v1"
        f"/workspaces/{workspace_id}"
        f"/items/{item_id}"
        f"/identities/default/assign?beta=true"
    )

    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json"
    }

    body = {
        "assignmentType": "Caller"
    }

    response = requests.post(endpoint, headers=headers, json=body)
    response.raise_for_status()

    # If synchronous success (unlikely but handle it)
    if response.status_code == 200:
        print(f"Identity associated immediately for item {item_id}.")
        return response.json()

    # 202 Accepted - poll the Location header
    location_url = response.headers.get("Location")
    if not location_url:
        raise RuntimeError(
            "Received 202 but no Location header for LRO polling."
        )

    print(f"Identity assignment started. Polling LRO for item {item_id}...")

    deadline = time.time() + poll_timeout_seconds

    while time.time() < deadline:
        poll_response = requests.get(location_url, headers={
            "Authorization": f"Bearer {access_token}"
        })
        poll_response.raise_for_status()

        poll_data = poll_response.json()
        status = poll_data.get("status")
        print(f"LRO status: {status}")

        if status == "Succeeded":
            print(f"Identity successfully associated for item {item_id}.")
            return poll_data

        if status in {"Failed", "Error", "Canceled"}:
            raise RuntimeError(
                f"Identity assignment failed with status: {status}. "
                f"Response: {poll_data}"
            )

        time.sleep(poll_interval_seconds)

    raise TimeoutError(
        f"Timed out after {poll_timeout_seconds}s waiting for identity "
        f"assignment on item {item_id}."
    )

# --------------------------------------------------------
#  Get item identity function
# --------------------------------------------------------
def get_item_identity(access_token: str, workspace_id: str, item_id: str) -> dict:
    """
    Retrieve the default identity information for a Fabric item.

    :param access_token: OAuth2 bearer token from Entra ID.
    :param workspace_id: GUID of the Fabric workspace.
    :param item_id: GUID of the Fabric item.
    :return: The item response dict including defaultIdentity info.
    """

    endpoint = (
        f"https://api.fabric.microsoft.com/v1"
        f"/workspaces/{workspace_id}"
        f"/items/{item_id}?include=defaultIdentity"
    )

    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json"
    }

    response = requests.get(endpoint, headers=headers)
    response.raise_for_status()

    data = response.json()

    identity_info = data.get("defaultIdentity", {})
    print(f"Item: {data.get('displayName')} ({item_id})")
    print(f"Identity: {identity_info}")

    return data
```

```python
# --------------------------------------------------------
#  LRO polling helper
# --------------------------------------------------------
def _poll_lro(
        access_token: str
      , location_url: str
      , poll_timeout_seconds: int = 120
      , poll_interval_seconds: int = 5
) -> dict:
    """
    Poll a Fabric long-running operation until it succeeds or times out.

    :param access_token: OAuth2 bearer token from Entra ID.
    :param location_url: Operation URL from the 202 Location header.
    :param poll_timeout_seconds: Maximum seconds to wait for the LRO.
    :param poll_interval_seconds: Seconds between polling attempts.
    :return: The final operation response body as a dict.
    :raises RuntimeError: If the operation ends Failed, Error, or Canceled.
    :raises TimeoutError: If the operation does not finish in time.
    """

    headers = {"Authorization": f"Bearer {access_token}"}
    deadline = time.time() + poll_timeout_seconds

    while time.time() < deadline:
        poll_response = requests.get(location_url, headers=headers)
        poll_response.raise_for_status()

        poll_data = poll_response.json()
        status = poll_data.get("status")
        print(f"LRO status: {status}")

        if status == "Succeeded":
            return poll_data

        if status in {"Failed", "Error", "Canceled"}:
            raise RuntimeError(
                f"Operation failed with status: {status}. Response: {poll_data}"
            )

        time.sleep(poll_interval_seconds)

    raise TimeoutError(
        f"Timed out after {poll_timeout_seconds}s waiting for operation at "
        f"{location_url}."
    )

# --------------------------------------------------------
#  Pipeline takeover function
# --------------------------------------------------------
def takeover_pipeline(
        access_token: str
      , workspace_id: str
      , item_id: str
) -> None:
    """
    Take over a Data Pipeline as the calling service principal.

    Pipelines have no associated-identity or ownership-takeover API -
    they execute under the identity recorded in LastModifiedBy. Re-saving
    the definition with the SP token flips LastModifiedBy to the SP
    without changing pipeline content.

    After takeover, runs execute as the SP, so every connection used by
    the pipeline's activities must be shared with the SP. Any later
    portal save by a user flips LastModifiedBy back to that user.

    :param access_token: OAuth2 bearer token from Entra ID (Fabric API audience).
    :param workspace_id: GUID of the Fabric workspace.
    :param item_id: GUID of the Data Pipeline item.
    :raises RuntimeError: If a 202 response has no Location header or the LRO fails.
    """

    base_url = (
        f"https://api.fabric.microsoft.com/v1"
        f"/workspaces/{workspace_id}"
        f"/dataPipelines/{item_id}"
    )

    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json"
    }

    # ----  Get the current definition (may return 202 + LRO)  ----
    response = requests.post(f"{base_url}/getDefinition", headers=headers)
    response.raise_for_status()

    if response.status_code == 200:
        definition = response.json()["definition"]
    else:
        location_url = response.headers.get("Location")
        if not location_url:
            raise RuntimeError(
                "Received 202 but no Location header for LRO polling."
            )

        print(f"getDefinition running. Polling LRO for pipeline {item_id}...")
        _poll_lro(access_token, location_url)

        result_response = requests.get(
            f"{location_url}/result"
            , headers={"Authorization": f"Bearer {access_token}"}
        )
        result_response.raise_for_status()
        definition = result_response.json()["definition"]

    # ----  Re-save the definition unchanged, minus .platform  ----
    # .platform must never be sent back on updateDefinition; keeping the
    # remaining parts (pipeline-content.json, .schedules) preserves the
    # pipeline content and schedules exactly.
    parts = [p for p in definition["parts"] if p["path"] != ".platform"]

    response = requests.post(
        f"{base_url}/updateDefinition"
        , headers=headers
        , json={"definition": {"parts": parts}}
    )
    response.raise_for_status()

    if response.status_code == 202:
        location_url = response.headers.get("Location")
        if not location_url:
            raise RuntimeError(
                "Received 202 but no Location header for LRO polling."
            )

        print(f"updateDefinition running. Polling LRO for pipeline {item_id}...")
        _poll_lro(access_token, location_url)

    print(f"Pipeline {item_id} taken over - LastModifiedBy is now the SP.")
```
